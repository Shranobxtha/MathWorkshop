```python
import numpy as np
import matplotlib.pyplot as plt
# Generate 100 equidistant point between -5 to 5 for x
x = np.linspace(-5,5,100)

## Calculate y values for Equation 1 and 2
y1 = (5-x)/3
y2 = (5+x)/2
y3 = 3-x

## Generate a figure frame for plotting the lines
fig, ax = plt.subplots()

## Plot the two lines
ax.plot(x,y1, color='r')
ax.plot(x,y2, color='g')
ax.plot(x,y3, color='b')

## To decorate the plot, we insert equations.
ax.text(0.5,0.6, '$x+3y = 5$', color='r')
ax.text(-3,0.5,'$x-2y = -5$', color='g')
ax.text(1.2,2, '$x+y = 3$', color='b')

## Set the limits of the X & Y axes.
ax.set_xlim(-4,4)
ax.set_ylim(-2,6)

## Set property of X and Y axes, e.g., color and linewidth
ax.axvline(color='k',linewidth = 1)
ax.axhline(color='k',linewidth = 1)

## This option specifies the ticks based the list of numbers provided.
ax.set_xticks(list(range(-4,5)))
ax.set_aspect('equal')
ax.grid(True,ls=':')
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
A = np.array([[1, 3],
              [1, -2],
              [1, 1]])

B = np.array([5, -5, 3])
solution, residuals, rank, s = np.linalg.lstsq(A, B, rcond=None)
print("Point of intersection is (x, y) = ",solution)
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Generate 100 equidistant point between -5 to 5 for x
x = np.linspace(-5,5,100)

## Calculate y values for Equation 1 and 2
y1 = (3-2*x)/3
y2 = (6-4*x)/6
y3 = (3*x-1)/2

## Generate a figure frame for plotting the lines
fig, ax = plt.subplots()

## Plot the two lines
ax.plot(x,y1, color='r')
ax.plot(x,y2, color='g')
ax.plot(x,y3, color='b')

## To decorate the plot, we insert equations.
ax.text(-3,1, '$2x+3y = 3$', color='r')
ax.text(-3,0.5,'$4x+6y = 6$', color='g')
ax.text(2.2,2, '$3x-2y = 1$', color='b')

## Set the limits of the X & Y axes.
ax.set_xlim(-4,4)
ax.set_ylim(-2,6)

## Set property of X and Y axes, e.g., color and linewidth
ax.axvline(color='k',linewidth = 1)
ax.axhline(color='k',linewidth = 1)

## This option specifies the ticks based the list of numbers provided.
ax.set_xticks(list(range(-4,5)))
ax.set_aspect('equal')
ax.grid(True,ls=':')
plt.xlabel("X")
plt.ylabel("Y")
plt.show()
A = np.array([[2, 3],
              [3, -2]])

B = np.array([3, 1])
print("Point of intersection is (x, y) = ",np.linalg.solve(A,B))
```

```python
#3
import numpy as np
matrix = np.random.randint(1, 21, size=(4, 4))
print("4x4 Matrix:")
print(matrix)

while True:
    user_input = input("\nEnter a number to search for: ")
    try:
        search_number = float(user_input)
        break
    except ValueError:
        print("Invalid input. Please enter a numeric value.")

count = np.sum(matrix == search_number)

if count > 0:
    print(f"The number {search_number} is in the matrix.")
    print(f"It appears {count} time(s).")
else:
    print(f"The number {search_number} was not found in the matrix.")
```

```python
#1
import numpy as np
A = np.random.randint(1, 10, (5, 5))
rank = np.linalg.matrix_rank(A)
is_invertible_rank = (rank == 5)
det = np.linalg.det(A)
is_invertible_det = not np.isclose(det, 0)
print("Matrix A:")
print(A)
print("-" * 30)
print(f"Rank: {rank}")
print(f"Invertible (check via Rank): {is_invertible_rank}")
print(f"Determinant: {det:.4f}")
print(f"Invertible (check via Determinant): {is_invertible_det}")
```

```python
#2
import numpy as np
A = np.random.randint(1, 10, (5, 5))
x = np.random.randint(1, 10, (5, 1))
b = A @ x
x_solved = np.linalg.solve(A, b)
print("Matrix A:")
print(A)
print()
print("Original Vector x:")
print(x)
print("\nSolved Vector x_solved:")
print(x_solved)
print("\nVerification (Original == Solved):", np.allclose(x, x_solved))
```

```python
import numpy as np
import matplotlib.pyplot as plt

x_points = np.array([-3, -2, -1, 0, 1, 2, 3, 4, 5, 7])
y_points = np.array([-7.6, 1, 4.8, 5, 2.8, -0.6, -4, -6.2, -6, 6.4])

A = np.vstack([x_points**3, x_points**2, x_points, np.ones(len(x_points))]).T

coefficients, residuals, rank, s = np.linalg.lstsq(A, y_points, rcond=None)
a, b, c, d = coefficients

print(f"Polynomial: {a:.2f}x^3 + {b:.2f}x^2 + {c:.2f}x + {d:.2f}")

x_curve = np.linspace(-4, 8, 100)
y_curve = a*x_curve**3 + b*x_curve**2 + c*x_curve + d

plt.scatter(x_points, y_points, color='red', label='Data Points')
plt.plot(x_curve, y_curve, label='Fitted Polynomial')
plt.axhline(0, color='black', linewidth=1)
plt.axvline(0, color='black', linewidth=1)
plt.legend()
plt.grid(True, ls=':')
plt.show()
```
